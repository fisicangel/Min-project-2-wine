## 3. Target Definition and Binary Encoding

The original `quality` variable contains numerical scores ranging from 3 to 9. For this classification project, the scores are converted into two classes:

- `0` = standard-quality wine, with a quality score below 7
- `1` = high-quality wine, with a quality score of 7 or higher

The threshold of 7 was selected to distinguish higher-rated wines from the more common scores of 5 and 6. This is a project-defined classification threshold and should not be interpreted as a universal wine-industry standard.

The Boolean results are converted into binary integers using `.astype(int)`. This is rule-based binary target encoding rather than One-Hot Encoding or Label Encoding.
